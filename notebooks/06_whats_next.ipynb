# 06. What is next: the 15th Five Year Plan, 2026 to 2030

The 15th Five Year Plan was adopted on March 12, 2026, by the National People's Congress, at the closing session of the fourth session of the 14th Congress. It has 20 main indicators across five dimensions (economic development, innovation, livelihood and well-being, green development, and security and resilience), 18 parts, 62 chapters, and 109 major projects. It also comes with a new law on national development planning, setting rules for how future plans are made.

## The headline goal

The core objective is to lay a solid foundation for doubling GDP per person by 2035 compared with 2020. That needs average growth of about 4.17 percent a year over the decade, according to EIAS analysis. That is slower than the 14th plan's 5.4 percent, and slower than any plan period since the 1990s, which tells you how much the growth model has changed.

## The 2026 starting point

The annual growth target for 2026 is 4.5 to 5.0 percent, described in secondary analysis as the lowest since 1991. Early data, from InsightBridge analysis citing NBS figures, shows GDP up 4.6 percent in the first three quarters of 2026, surveyed urban unemployment at 5.2 percent in September 2026, and youth unemployment (ages 16 to 24, excluding students) at 18.0 percent in September 2026. Per person disposable income in the first three quarters of 2026 was 34,230 yuan, up 5.3 percent nominal and 4.4 percent real. These are secondary analysis figures citing NBS, not primary releases, and they are labeled that way.

```python
import matplotlib.pyplot as plt
import pandas as pd

plt.rcParams["figure.figsize"] = (10, 5.5)
plt.rcParams["axes.titlesize"] = 13
plt.rcParams["axes.labelsize"] = 11
PROC = "../data/processed"
CHARTS = "../charts"

t15 = pd.read_csv(f"{PROC}/plan_15th_targets.csv")
print(t15[["indicator", "target_2030", "type"]].to_string(index=False))
```

```python

fig, ax = plt.subplots(figsize=(10, 6))
rows = [
    ("Digital economy share of GDP", 10.5, 12.5),
    ("Patents per 10,000 people", 16, 22),
    ("Working-age education, years", 11.3, 11.7),
    ("Life expectancy, years", 79.0, 80.0),
    ("Forest coverage", 25.0, 25.8),
    ("Non-fossil share of energy", 21.7, 25.0),
]
labels = [r[0] for r in rows]
v2025 = [r[1] for r in rows]
v2030 = [r[2] for r in rows]
y = range(len(rows))
ax.barh([i + 0.2 for i in y], v2025, height=0.4, label="2025 actual (or latest)", color="#969696")
ax.barh([i - 0.2 for i in y], v2030, height=0.4, label="2030 target", color="#2c7fb8")
ax.set_yticks(list(y))
ax.set_yticklabels(labels)
ax.set_title("15th plan: where China starts vs where it aims (different units, one chart)")
ax.legend()
ax.grid(True, axis="x", alpha=0.3)
plt.savefig(f"{CHARTS}/06_targets_2030.png", dpi=150, bbox_inches="tight")
print("saved 06_targets_2030.png")
print("Note: education 2025 value 11.3 is the 14th plan target, used as the start point")
print("because no verified 2025 actual was found. See notebook 03.")
```

## The targets, and where the sources disagree

- Research spending growth is reconfirmed at a minimum of 7 percent a year.
- Core digital economy value added is targeted at 12.5 percent of GDP by 2030, up from 10.5 percent in 2024.
- High-value invention patents are targeted at over 22 per 10,000 people by 2030, up from 16 in 2025.
- Working-age education is targeted at 11.7 years by 2030, binding, up from the 14th plan target of 11.3 years.
- Life expectancy is targeted at 80 years by 2030, up from 79.0 years (official, 2024).
- Carbon intensity is targeted to fall about 17 percent over the plan, binding. One summary table brackets 17.7 percent. Both figures are shown, not settled.
- Non-fossil energy share is targeted at 25 percent by 2030, binding, in the ICBC and Fidelity summaries. The iGDP summary cites about 27 percent from Government Work Report proposals. Both figures are shown, not settled.
- PM2.5 is targeted at under 27 micrograms per cubic meter by 2030, binding, down from 28 in 2025.
- Forest coverage is targeted at 25.8 percent by 2030, binding, up from past 25 percent in 2025.
- Grain production capacity is targeted at 725 million tons by 2030, binding, up from over 650 million tons in the 14th plan.
- Energy production capacity is targeted at 5.8 billion tons of coal equivalent by 2030, up from over 4.6 billion tons.
- 7 of the 20 indicators cover livelihood and well-being: education, health, elderly care, and childcare. That is the plan's biggest shift in emphasis.

## What changed in tone

Two small details say a lot. Artificial intelligence is mentioned 52 times in the 15th plan text, against 11 times in the 14th, according to South China Morning Post analysis cited in secondary summaries. And dual circulation, the slogan of the 14th plan, appears only once. The Made in China 2025 goal of 70 percent chip self-sufficiency, described in secondary analysis as largely missed, is dropped rather than repeated. The plan is more about technology, services, and living standards than about raw speed.

## What to watch

1. Growth versus the 4.17 percent decade average. If growth falls below about 4 percent for several years, the 2035 doubling goal slips out of reach.
2. Carbon intensity. The 14th plan missed its carbon target. The 15th plan's carbon target is binding, so a second miss would be harder to explain.
3. Youth unemployment. At 18.0 percent in September 2026 (excluding students), it is the most politically sensitive number in the plan.
4. Births and the population. The population fell for a fourth straight year in 2025. Childcare subsidies started in January 2026 (3,600 yuan a year per child under 3). Watch whether births stabilize.
5. Property and debt. Property sales are down 54 percent from 2021, and total non-financial debt is near 296 percent of GDP. The plan's growth math assumes these do not blow up.
6. The non-fossil and carbon target discrepancy. If the adopted plan text settles on 27 percent rather than 25 percent for non-fossil energy, or 17.7 percent rather than 17 percent for carbon, this notebook should be updated. Both figures are kept until the plan text is checked.

## How this project ends
