# 01. How planning works, in plain words

## The basic idea

Every five years, China writes a national development plan. Think of it as a to do list for the whole country, with numbers attached. The plan says what the government wants to build, protect, and improve over the next five years, and how it will know whether it succeeded.

The system started in 1953, copying the Soviet Union. It has changed a lot since then. Early plans were strict orders: factories were told exactly what to produce. Modern plans are more like guidelines with a few hard promises mixed in. Markets now do most of the day to day deciding. The plan sets the direction.

## The five year cycle, step by step

1. Midway through the current plan, work starts on the next one. Drafting takes two to three years.
2. The Communist Party Central Committee agrees on general guidelines in the autumn before the plan starts.
3. The State Council (the cabinet) writes the detailed plan.
4. The National People's Congress (the legislature) approves it the following March.
5. Ministries, provinces, and cities write their own detailed action plans to match.
6. Progress is reviewed, and the cycle starts again.

So a plan is not one document written in a week. It is a long coordination exercise across the whole government.

## Two kinds of targets: binding and indicative

This distinction matters, and it explains a lot of the scorecard in notebook 03.

- A binding target is a promise. The government says it must be hit, and local officials are judged on it. Binding targets are used mostly for the environment, resources, and social welfare. Example: cut energy intensity by 13.5 percent.
- An indicative target is a goal, not an order. It says where the government wants to go, but missing it is not treated as a broken promise. Example: raise the urbanization rate to 65 percent.

Since the 11th plan (2006 to 2010), binding targets have been written into how local officials are evaluated. That is why environmental targets get more attention than they used to.

## What is a five year plan not?

It is not a budget. It does not list every project and its price. It is not a law that punishes companies for missing a number. And since the 14th plan, it does not even set a number for GDP growth. The 14th plan said growth should stay in a reasonable range, without naming the range. That was a first.

## The long-range objectives through 2035

On top of the five year plans, China sets longer goals. The current long-range objectives run to 2035. They describe a country that is much richer per person, more innovative, greener, and with a stronger safety net. The 15th plan (2026 to 2030) is supposed to lay the foundation for doubling GDP per person by 2035 compared with 2020. Notebook 06 explains what that requires.

## The plans so far

The chart below shows every plan period from 1953 to 2030. Two gaps are real, not mistakes. No five year plan covered 1963 to 1965, because the economy was too dislocated after the Great Leap Forward to plan. And the 5th plan period (1976 to 1980) started with a ten year outline that was later rewritten when reform began in December 1978.

```python
import matplotlib.pyplot as plt
import pandas as pd

plt.rcParams["figure.figsize"] = (10, 5.5)
plt.rcParams["axes.titlesize"] = 13
plt.rcParams["axes.labelsize"] = 11
PROC = "../data/processed"
CHARTS = "../charts"

import matplotlib.patches as mpatches

plans = [
    ("1st", 1953, 1957), ("2nd", 1958, 1962), ("3rd", 1966, 1970),
    ("4th", 1971, 1975), ("5th", 1976, 1980), ("6th", 1981, 1985),
    ("7th", 1986, 1990), ("8th", 1991, 1995), ("9th", 1996, 2000),
    ("10th", 2001, 2005), ("11th", 2006, 2010), ("12th", 2011, 2015),
    ("13th", 2016, 2020), ("14th", 2021, 2025), ("15th", 2026, 2030),
]
fig, ax = plt.subplots(figsize=(11, 5))
for i, (name, start, end) in enumerate(plans):
    color = "#2c7fb8" if i < 5 else ("#41b6c4" if i < 10 else "#a1dab4")
    ax.barh(0, end - start + 1, left=start, height=0.5, color=color)
    ax.text((start + end) / 2, 0, name, ha="center", va="center",
            fontsize=8, color="black")
ax.set_xlim(1952, 2031)
ax.set_ylim(-0.6, 0.6)
ax.set_yticks([])
ax.set_xlabel("Year")
ax.set_title("China's five year plans, 1953 to 2030 (no plan covered 1963 to 1965)")
ax.legend(handles=[mpatches.Patch(color="#2c7fb8", label="Mao era plans"),
                   mpatches.Patch(color="#41b6c4", label="Reform era plans"),
                   mpatches.Patch(color="#a1dab4", label="Recent plans")],
          loc="lower center", bbox_to_anchor=(0.5, -0.45), ncol=3, frameon=False)
plt.tight_layout()
plt.savefig(f"{CHARTS}/01_plan_timeline.png", dpi=150, bbox_inches="tight")
print("saved 01_plan_timeline.png")
```

## Words you will meet in this project

- GDP: the total value of everything produced in a country in a year. GDP per person divides that by the population. It is a rough measure of average income, not of happiness or fairness.
- Energy intensity: how much energy is used to produce one unit of GDP. If it falls, the economy is getting more output from each ton of coal or kilowatt hour.
- Carbon intensity: how much carbon dioxide is emitted per unit of GDP. It can fall even while total emissions rise, if GDP grows faster.
- Non-fossil energy: energy from sources other than coal, oil, and gas. That means hydro, nuclear, wind, solar, and others.
- Urbanization rate: the share of the population living in cities and towns, counting people who live there permanently, not just people with a city household registration.
- Hukou: the household registration system. It ties a person's access to local schools, health care, and pensions to the place they are registered, which is often their birthplace. Migrant workers living in cities without a local hukou are a big policy issue.
- Purchasing power parity (PPP): a way to compare economies that adjusts for price differences. A haircut costs less in China than in the United States, so PPP makes China's economy look bigger than market exchange rates do. This project uses market rates unless it says PPP.
