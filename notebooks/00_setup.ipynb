# 00. Setup: what this project is, and how to check it

Welcome. This project answers one question in plain words: how does China plan its economy in five year blocks, did the last plan hit its targets, how do the provinces differ, who lives where, and what comes next?

China does not hold competitive presidential elections, so there is no elections chapter here. This is a planning, demographics, and macroeconomics project. If you want elections data, look at the companion projects on the United States and Brazil.

## How to read this project

Start at notebook 01 if you know nothing about five year plans. It explains every idea before using it. Notebook 02 tells the long story, from the first plan in 1953 to today, including the failures. Notebook 03 scores the 14th Five Year Plan (2021 to 2025) target by target. Notebook 04 compares the provinces. Notebook 05 covers people and places. Notebook 06 looks at the 15th Five Year Plan (2026 to 2030).

## Rules this project follows

- Every number names its source. Official figures are labeled official. Independent figures are labeled independent.
- Where sources disagree, both figures are shown and no side is taken.
- Where a figure could not be verified, the project says so and does not guess.
- No dataset, map, or test is faked. Cuts are listed in BUILD_REPORT.md.

## Data sources at a glance

- World Bank, World Development Indicators (no key needed). Long run series for GDP, population, urbanization, life expectancy, fertility, poverty, trade, energy, research spending, and age structure. These are the independent cross-check series.
- National Bureau of Statistics of China (NBS), 2025 Statistical Communique, February 28, 2026. The official 2025 figures: GDP, income, urbanization, energy, environment, jobs.
- Government Work Report, delivered March 5, 2026. The official verdict on the 14th plan and the 15th plan headline targets.
- Ministry of Ecology and Environment reporting, via Xinhua, January and March 2026. Air and water figures.
- iGDP China Carbon Neutrality Tracker newsletter, January to February 2026 issue. Full period energy and carbon intensity figures.
- Seventh National Population Census, 2020, results released May 11, 2021. Population, ethnic composition, and provincial populations.
- NBS provincial data via Wikipedia compilations of NBS tables. Provincial GDP for 2024 (revised) and 2025 (preliminary), and mid-2022 provincial populations with urban and rural splits.
- Pew Research Center (2023), CIA World Factbook (2021), and China Social Survey (2010) for religion. Three different methods, three different answers, all shown.
- Plan documents and secondary summaries for the 15th plan (adopted March 12, 2026), including EIAS analysis and ICBC and Fidelity summary tables. Where those tables disagree, both figures are kept.

Full citations, file by file, are in data/raw/PROVENANCE.md.

## Verification checks

The next cell runs the checks this project promises. It confirms the World Bank series load, the provincial table sums to about the national total (a small gap is normal, see notebook 04), the 14th plan scorecard has 20 rows, and the ethnic shares add to about 100 percent. If any check fails, the build is not finished.

```python

prov = pd.read_csv(f"{PROC}/provinces.csv")
score = pd.read_csv(f"{PROC}/plan_14th_scorecard.csv")
eth = pd.read_csv(f"{PROC}/ethnic_2020.csv")
print("provinces:", len(prov), "(31 mainland divisions expected)")
print("provincial 2024 GDP sum, million CNY:", prov.gdp_2024_m_cny.sum())
print("national 2024 GDP, million CNY: 134806620")
print("gap:", round((1 - prov.gdp_2024_m_cny.sum() / 134806620) * 100, 2), "percent")
print("14th plan scorecard rows:", len(score), "(20 expected)")
print("scorecard status counts:"); print(score.status.value_counts().to_dict())
print("ethnic shares sum:", round(eth.share_pct.sum(), 2), "percent")
print("ALL CHECKS DONE")
```
